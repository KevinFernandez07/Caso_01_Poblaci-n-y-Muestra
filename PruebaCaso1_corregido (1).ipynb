# Objetivo
###### Realizar una prueba de cargar datos de registros de las personas y su comodidad o ciclo de sueño

# Descripción
Este trabajo analiza los patrones de sueño de 50 estudiantes de nivel superior (la población). Cada registro contiene: horas de sueño, calidad del sueño (1-10) y nivel de descanso (Mala, Regular, Buena). A partir de esa población se extrae una muestra de 30 estudiantes, y se comparan los parámetros de la población contra los estadísticos de la muestra.

Variable: Característica o atributo que se mide u observa en cada estudiante (por ejemplo, horas de sueño, calidad del sueño).

Media: Promedio aritmético de los valores; se obtiene sumando todos los datos y dividiendo entre el número total de observaciones.

Mediana: Valor central de los datos cuando están ordenados de menor a mayor; divide la muestra en dos partes iguales.

Moda: Valor o categoría que aparece con mayor frecuencia en el conjunto de datos. Es la única medida de tendencia central que también aplica a variables categóricas.

Mínimo: Valor más pequeño observado en la variable.

Máximo: Valor más grande observado en la variable.

Rango: Diferencia entre el máximo y el mínimo; indica la amplitud de variación de los datos.

Desviación (estándar): Medida que indica cuánto se dispersan, en promedio, los datos respecto a la media. Solo aplica a variables numéricas.

Frecuencia: Para variables categóricas (como Nivel_descanso), en vez de media/mediana/desviación estándar se calculan la moda (categoría más común) y la frecuencia o porcentaje de cada categoría.

# Desarrollo

## Estadísticos de la muestra
Estadísticos descriptivos de las variables numéricas

```python
muestra.describe()
```

Estadísticos descriptivos de la variable categórica (Nivel_descanso) en la muestra

```python
print("Frecuencia de cada categoría en la muestra:")
print(muestra['Nivel_descanso'].value_counts())
print()
print("Moda (categoría más frecuente):", muestra['Nivel_descanso'].mode()[0])
```

# Interpretación

**¿Cuál es el contexto de los datos y cuáles son las variables de interés?**

Muestra simulada de 50 estudiantes de nivel superior sobre sus hábitos de sueño. Las variables de interés son: `ID_estudiante` (identificador), `Horas_sueno` (numérica), `Calidad_sueno_1_a_10` (numérica) y `Nivel_descanso` (categórica: Mala, Regular, Buena).

**¿Qué parámetros de la población y estadísticos de la muestra son calculados con las variables numéricas?**

Para `Horas_sueno` y `Calidad_sueno_1_a_10` se calculan, tanto en la población como en la muestra: media, mediana, moda, mínimo, máximo, rango y desviación estándar. Cuando estos valores vienen de los 50 registros completos se llaman *parámetros* (población); cuando vienen de los 30 registros de la muestra se llaman *estadísticos* (muestra).

**¿De las variables categóricas qué parámetros tanto de la población como de la muestra se pueden calcular?**

Para una variable categórica como `Nivel_descanso` no tiene sentido calcular media, mediana o desviación estándar (no se puede promediar una categoría de texto). Lo que sí se puede calcular, tanto en la población como en la muestra, es la **moda** (la categoría más frecuente) y la **frecuencia o porcentaje** de cada categoría. En este caso, en la población: Buena=27, Regular=22, Mala=1 (moda = "Buena"); en la muestra: Buena=19, Regular=11, Mala=0 (moda = "Buena" también).

**De los parámetros calculados de la población y de los estadísticos determinados de la muestra de las variables de interés de tipo numérica, ¿son similares o existe diferencia?, ¿por qué?**

Son bastante parecidos. La media poblacional de `Horas_sueno` es 6.72 y la de la muestra es 6.92 (diferencia de apenas 0.2 horas); la desviación estándar es casi idéntica (0.97 vs 0.96). Para `Calidad_sueno_1_a_10` la media poblacional es 7.54 y la de la muestra 7.80, también muy cercanas. Esto pasa porque la muestra de 30 sobre una población de 50 es una proporción grande (60%), así que captura bien el comportamiento general del grupo. Sí hay una diferencia que vale la pena señalar con honestidad: en la variable categórica, la muestra se quedó sin ningún caso "Mala" (0 de 30), mientras que la población sí tenía uno (1 de 50, apenas 2%). Esto no es un error — es esperable, porque al ser una categoría tan rara, es fácil que una muestra aleatoria no la incluya. Es un buen ejemplo de que, aunque la muestra sea representativa en general, las categorías muy poco frecuentes en la población pueden no aparecer en una muestra más pequeña.

**¿Qué te deja el caso?**

Que con la estadística descriptiva podemos resumir y entender mejor los datos de una muestra, y que esos números nos dan una idea de cómo es la población completa. También me ayuda a ver la diferencia entre parámetro (de toda la población, que normalmente no conocemos) y estadístico (de la muestra, que sí podemos calcular), y a distinguir qué medidas aplican a variables numéricas (media, desviación estándar, etc.) y cuáles a variables categóricas (moda, frecuencia). Y sobre todo, que para que las conclusiones sirvan, la muestra tiene que representar bien a la población — aunque incluso una buena muestra puede no capturar categorías muy poco frecuentes, como pasó aquí con "Mala".
