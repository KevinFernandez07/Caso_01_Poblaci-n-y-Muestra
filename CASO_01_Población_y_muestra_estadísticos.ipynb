# Objetivo

Cargar, analizar e intrepretar estadísticamente datos de  registros de estudiantes.



# Descripción

El contexto está relacioando con la educación y el conjunto de datos contiene registros y variables de estudiantes.

Las variables de interés:

* *Horas_Estudio_Semana* es una variable cuantitativa que define la cantidad de hora que estudia el alumno a la semana.
* *Asistencia_Porcentaje* mide la cantidad de asistencia promedio a clase del alumno
* *Modalidad*	Variable cualitativa que define el tipo de modalidad en que estudia el alumno.
* *Rendimiento*. Es un valor que define el rendimiento académico del alumno, la variable es cualitativa.


## Cargar librerías


```python
import pandas as pd # Cargar la librería que procesa y trata conjuntos de datos
```

## Cargar datos


```python
# Ruta local en sesión google collab
ruta = "sample_data/datos_educacion_100.csv"

# ruta en internet en github.com
# ruta = "https://raw.githubusercontent.com/rpizarrog/probabilidad-y-estad-stica/refs/heads/master/2026/datos/datos_educacion_100.csv"


datos = pd.read_csv(ruta)
datos


```

# Desarrollo

## Parámetros de población

La población son todos los datos cargados.
Los parámetros son cálculos estadísticos de la población.

Parámetros de variables numéricas



```python
datos.describe()
```

Parámetros de variables categóricas


```python
datos.describe(include="object")
```

## Extraer una muestra

Se extrae una muestar de la población de $n=30$.


```python
muestra = datos.sample(n=30)
muestra
```

## Estadisticos de la muestra

Estadisticos descruiptivos de las variables numericas



```python
muestra.describe()
```

Estadísticos descriptivos de las variables categóricas

```python
muestra.describe(include="object") # Las variables categóricas
```
